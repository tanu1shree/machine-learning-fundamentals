```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Finance_Analytics_Dataset.csv")
df
```

1. Show the number of transactions in each category using a count plot.

```python
sns.countplot(x="Category", data=df, hue="Category", palette="Set1", legend=False)
plt.title("Total Number of Transactions in Each Category")
plt.show()
```

2. Show the number of transactions for each payment mode using a count plot.

```python
sns.countplot(x="PaymentMode", data=df, hue="PaymentMode", palette="Set1", legend=False)
plt.title("Total Number of Transactions in Each Payment Mode")
plt.show()
```

3. Show the number of transaction by transaction type (Debit and Credit) using a count plot.

```python
sns.countplot(x="TransactionType", data=df, hue="TransactionType", palette="Set1", legend=False)
plt.title("Number of Transactions by Transaction Type")
plt.show()
```

4. Debit vs Credit Amount Comparison (Bar Plot)

```python
sns.barplot(x="TransactionType", y="Amount", data=df, hue="TransactionType", palette="Set1", legend=False)
plt.title("Debit vs Credit Amount Comparison")
plt.ylabel("Transaction Amount")
plt.show()
```

5. Spending Pattern by Custom Segment (Bar Plot)

```python
sns.barplot(x="CustomerSegment", y="Amount", data=df, hue="CustomerSegment", palette="Set1", legend=False)
plt.title("Spending Pattern by Customer Segment")
plt.show()
```

6. Payment Mode Usage Distribution (Pie Chart)

```python
payment_counts = df["PaymentMode"].value_counts()
plt.pie(payment_counts, labels=payment_counts.index, autopct="%1.1f%%")
plt.title("Payment Mode Usage Distribution")
plt.show()
```

7. Boxplot of Transaction Amount by Category

```python
sns.boxplot(x="Amount", y="Category", data=df, hue="Category", palette="Set1", legend=False)
plt.show()
```
