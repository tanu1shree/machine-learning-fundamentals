# HOUSE PRICES ANALYSIS

Steps: 
1. import libraries
2. load data
3. handle missing values (preprocessing)
4. EDA
5. define x(independent) and y(dependent, the variable we aim to find)
6. model building, model -linear regression
7. training the model. train test split. model.fit(X_train, Y_train)
8. testing - model.predict(X_test)
9. evaluation, print evaluation metrics
10. squaring the error -> to deal with negative sign -> mean absolute error should be minimum... r squared core: how good the model works percentage (should be high)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv("house.csv")
df
```

```python
X = df['House Size'].to_frame()
Y = df['House Price']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, Y_train)
```

```python
Y_pred = model.predict(X_test)
Y_pred
```

```python
r2_score = r2_score(Y_test, Y_pred)
r2_score
# Print Evaluation metrics
print(f"R-squared Score:", r2_score)
```

```python
user_House_Size = int(input("Enter House Size to predict salary: "))
user_df = pd.DataFrame([[user_House_Size]], columns=['House Size'])
predicted_salary = model.predict(user_df)
print(f"The predicted House Price for House Size {user_House_Size} is:", predicted_salary)
```

```python
# Plot
plt.scatter(X, Y, color='blue', label='Data points') # All data points
plt.plot(X, model.predict(X), color='red', label='Regression Line') # Regression Line
plt.xlabel('House Size (sqft)')
plt.ylabel('House Price ($)')
plt.title('House Price vs Size')
plt.legend()
plt.show()
```
